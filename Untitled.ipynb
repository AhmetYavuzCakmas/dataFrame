```python
num_rows = dataFrame.shape[0]
print(f"toplam satır sayısı {num_rows}")
```

```python
dataFrame1 = pd.DataFrame(dataFrame)
ortalama_maas = dataFrame1['Salary'].mean()
print(f"ortalama maas: {ortalama_maas:.3f}")
```

```python
dataFrame2 = pd.DataFrame(dataFrame)
departmana_gore_ortalama_maas = dataFrame.groupby('Department')['Salary'].mean()
for department, avg_salary in departmana_gore_ortalama_maas.items():
    print(f"{department}: {avg_salary:.3f}")
```

```python
dataFrame3 = pd.DataFrame(dataFrame)
basliga_gore_ortalama_maas = dataFrame2.groupby('Title')['Salary'].mean()
print(basliga_gore_ortalama_maas.round(3))
```

```python
dataFrame4 = pd.DataFrame(dataFrame)
average_salary_by_title = dataFrame4.groupby('Title')['Salary'].mean()
junior_avg_salary = average_salary_by_title.get('Junior', 0) 
senior_avg_salary = average_salary_by_title[['Mid-Senior', 'C-level']].mean()
percent_increase = ((senior_avg_salary - junior_avg_salary) / junior_avg_salary) * 100 
print(f"Senior bir kişinin junior bir kişiye göre maaşı ortalama yüzde {percent_increase:.2f} daha fazladır.")
```
